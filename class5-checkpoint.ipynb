```python
marks = float(input(Enter marks: "))
if marks < 25:
    grade = "F"
elif marks >= 25 and marks < 45:
        grade = "E"
elif marks >= 45 and marks < 50:
      grade = "D"
elif marks >= 50 and marks < 60:
      grade = "C"
elif marks >= 60 and marks < 80:
      grade = "B"
else : # marks > 80
       grade = "A"
print("Tour grade is:" , grade)
```

```python
lst = list(map(int, input("Enter list elements seprated by space: ").split()))
print("Original list:" , lst)
largest = lst[0]
for num in lst:
    if num > largest:
        largest = num
        print("Largest number in the list is:" , largest)
```
